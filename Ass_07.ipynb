```python
# ==========================================
# ASSIGNMENT 1
# Undiscounted and Discounted Returns
# ==========================================

print("REINFORCEMENT LEARNING")
print("Undiscounted and Discounted Returns")
print("-" * 40)

# Number of time steps
n = int(input("Enter the number of time steps: "))

# Enter rewards
rewards = []

for i in range(n):
    r = float(input(f"Enter reward for time step {i + 1}: "))
    rewards.append(r)

# Undiscounted return
undiscounted_return = sum(rewards)

# Discount factor
gamma = float(input("Enter discount factor (gamma): "))

# Discounted return
discounted_return = 0

for t in range(n):
    discounted_return += rewards[t] * (gamma ** t)

# Display results
print("\n========== RESULTS ==========")
print("Rewards:", rewards)
print("Discount Factor (γ):", gamma)
print("Undiscounted Return:", undiscounted_return)
print("Discounted Return:", discounted_return)

```

```python
# ==========================================
# ASSIGNMENT 2
# 2 x 3 GRID WORLD MDP
# ==========================================

print("===================================")
print("       2 x 3 GRID WORLD MDP")
print("===================================")

# Grid:
#
#       S0      S1       G
#       ------------------
#       X       S2       S3
#
# S0 = Starting state
# G  = Goal state
# X  = Terminal/blocked state


class GridWorld:

    def __init__(self):

        # Environment name
        self.name = "grid_env"

        # Grid dimensions
        self.rows = 2
        self.cols = 3

        # State space
        self.states = ["S0", "S1", "G", "X", "S2", "S3"]

        # Action space
        self.actions = ["up", "down", "left", "right"]

        # Discount factor
        self.gamma = 0.9

        # State positions
        self.positions = {
            "S0": (0, 0),
            "S1": (0, 1),
            "G":  (0, 2),
            "X":  (1, 0),
            "S2": (1, 1),
            "S3": (1, 2)
        }

        # Reward function
        self.rewards = {
            "S0": -1,
            "S1": -1,
            "G": 10,
            "X": -10,
            "S2": -1,
            "S3": -1
        }

        # Initial state
        self.current_state = "S0"


    # --------------------------------------
    # Transition Function
    # --------------------------------------

    def transition(self, state, action):

        row, col = self.positions[state]

        if action == "up":
            new_row = row - 1
            new_col = col

        elif action == "down":
            new_row = row + 1
            new_col = col

        elif action == "left":
            new_row = row
            new_col = col - 1

        elif action == "right":
            new_row = row
            new_col = col + 1

        else:
            print("Invalid action!")
            return state

        # Check boundaries
        if new_row < 0 or new_row >= self.rows:
            return state

        if new_col < 0 or new_col >= self.cols:
            return state

        # Find next state
        for s, position in self.positions.items():

            if position == (new_row, new_col):
                return s

        return state


    # --------------------------------------
    # Reward Function
    # --------------------------------------

    def get_reward(self, state):
        return self.rewards[state]


    # --------------------------------------
    # Display Grid
    # --------------------------------------

    def display_grid(self):

        print("\nGrid World:")
        print("+-------+-------+-------+")
        print("|  S0   |  S1   |   G   |")
        print("+-------+-------+-------+")
        print("|   X   |  S2   |  S3   |")
        print("+-------+-------+-------+")


    # --------------------------------------
    # Take Action
    # --------------------------------------

    def step(self, action):

        next_state = self.transition(
            self.current_state,
            action
        )

        reward = self.get_reward(next_state)

        print("\nCurrent State :", self.current_state)
        print("Action        :", action)
        print("Next State    :", next_state)
        print("Reward        :", reward)

        self.current_state = next_state

        return next_state, reward


# ==========================================
# Create MDP
# ==========================================

grid_env = GridWorld()

# Display grid
grid_env.display_grid()

# Display state space
print("\nState Space:")
print(grid_env.states)

# Display action space
print("\nAction Space:")
print(grid_env.actions)

# Display discount factor
print("\nDiscount Factor:")
print(grid_env.gamma)

# Display initial state
print("\nInitial State:")
print(grid_env.current_state)


# ==========================================
# Take User Action
# ==========================================

action = input(
    "\nEnter action (up/down/left/right): "
).lower()

if action in grid_env.actions:

    grid_env.step(action)

else:

    print("Invalid action!")


print("\n===================================")
print("       MDP EXECUTION COMPLETED")
print("===================================")
```
