```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset (traffic_source.csv should have columns: Source, Visits)
data = pd.read_csv("traffic_source.csv")

# Display first few rows
print(data.head())

# Plot Bar Chart
plt.figure(figsize=(8,6))
bars = plt.bar(data["Source"], data["Visits"], color="skyblue", edgecolor="black")

# Add value labels on top of bars
for bar in bars:
    plt.text(bar.get_x() + bar.get_width()/2,
             bar.get_height() + 5,   # position above bar
             str(bar.get_height()),
             ha="center", fontsize=10)

# Chart details
plt.title("Website Traffic by Source", fontsize=14, weight="bold")
plt.xlabel("Traffic Source")
plt.ylabel("Number of Visits")
plt.grid(axis="y", linestyle="--", alpha=0.7)

plt.show()

# Find highest traffic source
top_source = data.loc[data["Visits"].idxmax()]
print(f"Highest Traffic Source: {top_source['Source']} with {top_source['Visits']} visits")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset (monthly_rainfall.csv should have columns: Month, Rainfall)
data = pd.read_csv("monthly_rainfall.csv")

# Define seasonal colors (Winter: blue, Summer: orange, Monsoon: green, Autumn: brown)
season_colors = {
    "Jan": "skyblue", "Feb": "skyblue", "Dec": "skyblue",       # Winter
    "Mar": "orange", "Apr": "orange", "May": "orange",          # Summer
    "Jun": "green", "Jul": "green", "Aug": "green", "Sep": "green",  # Monsoon
    "Oct": "peru", "Nov": "peru"                                # Autumn
}

colors = [season_colors[month] for month in data["Month"]]

# Plot bar chart
plt.figure(figsize=(10,6))
bars = plt.bar(data["Month"], data["Rainfall"], color=colors, edgecolor="black")

# Add value labels
for bar in bars:
    plt.text(bar.get_x() + bar.get_width()/2,
             bar.get_height() + 3,
             str(bar.get_height()),
             ha="center", fontsize=9)

# Chart details
plt.title("Average Monthly Rainfall", fontsize=14, weight="bold")
plt.xlabel("Month")
plt.ylabel("Rainfall (mm)")
plt.grid(axis="y", linestyle="--", alpha=0.6)

plt.show()

# Identify wettest & driest month
wettest = data.loc[data["Rainfall"].idxmax()]
driest = data.loc[data["Rainfall"].idxmin()]

print(f"Wettest Month: {wettest['Month']} with {wettest['Rainfall']} mm")
print(f"Driest Month: {driest['Month']} with {driest['Rainfall']} mm")
```
