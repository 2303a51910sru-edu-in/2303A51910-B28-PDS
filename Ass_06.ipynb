```python

# Assignment 1: Calculate State Value and Action Value
# Grid World (2 x 3)

# Environment details
gamma = 0.9
normal_reward = -1
goal_reward = 10
obstacle_reward = -5

# States and policy
states = ["S0", "S1", "S3", "S4", "G"]

policy = {
    "S0": "Right",
    "S1": "Down",
    "S3": "Right",
    "S4": "Up"
}

# State transition function
def get_next_state(state, action):

    transitions = {
        ("S0", "Right"): "S1",
        ("S0", "Down"): "X",
        ("S0", "Left"): "S0",
        ("S0", "Up"): "S0",

        ("S1", "Right"): "G",
        ("S1", "Down"): "S3",
        ("S1", "Left"): "S0",
        ("S1", "Up"): "S1",

        ("S3", "Right"): "S4",
        ("S3", "Down"): "S3",
        ("S3", "Left"): "X",
        ("S3", "Up"): "S1",

        ("S4", "Right"): "S4",
        ("S4", "Down"): "S4",
        ("S4", "Left"): "S3",
        ("S4", "Up"): "G"
    }

    return transitions.get((state, action), state)


# Reward function
def get_reward(next_state):
    if next_state == "G":
        return goal_reward
    elif next_state == "X":
        return obstacle_reward
    else:
        return normal_reward


# State Value Function
def state_value_function(state):

    if state == "G" or state == "X":
        return 0

    action = policy[state]
    next_state = get_next_state(state, action)
    reward = get_reward(next_state)

    if next_state == "G" or next_state == "X":
        return reward

    return reward + gamma * state_value_function(next_state)


# Action Value Function
def action_value_function(state, action):

    next_state = get_next_state(state, action)
    reward = get_reward(next_state)

    if next_state == "G" or next_state == "X":
        return reward

    return reward + gamma * state_value_function(next_state)


# Calculate State Value of S0
V_S0 = state_value_function("S0")

# Calculate Action Value of S0 for UP
Q_S0_Up = action_value_function("S0", "Up")

# Display results
print("State Value Function")
print("V(S0) =", round(V_S0, 4))

print("\nAction Value Function")
print("Q(S0, Up) =", round(Q_S0_Up, 4))
```
